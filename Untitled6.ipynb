```python
%%writefile app_rust.py
import streamlit as st

# 1. BASE DE DATOS DE RUST
database = {
    "Sheet Metal Door": {"Rocket": 2, "C4": 1, "Satchel": 4, "Explosive Ammo": 63},
    "Stone Wall": {"Rocket": 4, "C4": 2, "Satchel": 10, "Explosive Ammo": 185},
    "Sheet Metal Wall": {"Rocket": 8, "C4": 4, "Satchel": 23, "Explosive Ammo": 400},
    "Armored Wall": {"Rocket": 15, "C4": 8, "Satchel": 46, "Explosive Ammo": 799}
}

crafting_costs = {
    "Rocket": {"sulfur": 1400, "charcoal": 1950, "metal": 150},
    "C4": {"sulfur": 2200, "charcoal": 3000, "metal": 200},
    "Satchel": {"sulfur": 480, "charcoal": 720, "metal": 80},
    "Explosive Ammo": {"sulfur": 25, "charcoal": 30, "metal": 10}
}

# 2. DISEÑO DE LA PÁGINA WEB
st.set_page_config(page_title="Rust Raid Calculator", page_icon="💥", layout="centered")

st.title("💥 Ultimate Rust Raid Calculator 💥")
st.write("Configura tu plan de raideo para calcular los materiales exactos que necesitas farmear.")

st.divider()

# Columnas para los menús desplegables
col1, col2, col3 = st.columns(3)

with col1:
    structure = st.selectbox("Estructura Objetivo", list(database.keys()))

with col2:
    quantity = st.number_input("Cantidad", min_value=1, value=1, step=1)

with col3:
    explosive = st.selectbox("Explosivo a usar", list(crafting_costs.keys()))

st.divider()

# 3. LÓGICA DE CÁLCULO
ammo_needed = database[structure][explosive] * quantity

multiplier = ammo_needed
if explosive == "Explosive Ammo":
    multiplier = max(1, ammo_needed // 2)

total_sulfur = multiplier * crafting_costs[explosive]["sulfur"]
total_charcoal = multiplier * crafting_costs[explosive]["charcoal"]
total_metal = multiplier * crafting_costs[explosive]["metal"]

# 4. MOSTRAR RESULTADOS EN PANTALLA
st.subheader("📋 Reporte del Raideo")

# Tarjeta llamativa para la munición total
st.metric(label=f"Total de {explosive} necesarios", value=f"{ammo_needed:,}")

st.write("### ⛏️ Materiales Brutos para Fabricar Todo:")

# Diseñar cajitas visuales para cada material
m1, m2, m3 = st.columns(3)
with m1:
    st.info(f"**Azufre (Sulfur)**\n\n {total_sulfur:,} unidades")
with m2:
    st.success(f"**Carbón (Charcoal)**\n\n {total_charcoal:,} unidades")
with m3:
    st.error(f"**Metal (Frags)**\n\n {total_metal:,} unidades")
```
